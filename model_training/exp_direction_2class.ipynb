# Thử nghiệm: Direction 2 lớp (lên / xuống 8h) chọn hướng trước, touch lọc biến động sau

**Giả thuyết:** Direction chỉ nên đoán hướng (2 lớp) trên mọi giờ. Hướng được chọn trước, rồi touch lọc những giờ có biến động đủ để giao dịch.

**Nhãn Direction**, tính trên MỌI giờ, 8 nến kể từ open[t+1]:
| nhãn | lớp | điểm hướng |
|---|---|---|
| `3class` (hiện tại) | không chạm / chạm +b trước / chạm −b trước | p_up = 0.5 + 0.5·(P↑ − P↓) |
| `ret8` | close[t+8] > open[t+1] hay không | p_up = P(lên) |
| `tb_sign` | đã chạm: mức chạm trước; chưa chạm hoặc chạm cùng nến: dấu lợi suất cuối 8h | p_up = P(lên) |

**Thứ tự lọc:**
- `touch_first` (hiện tại): ngưỡng touch trên mọi giờ (14 ngày); ngưỡng Direction trên các giờ qua touch (90 ngày).
- `dir_first`: ngưỡng Direction trên mọi giờ (90 ngày, cân bằng 2 phía); ngưỡng touch trên các giờ 14 ngày trước đã qua Direction.

Mỗi tổ hợp nhãn × thứ tự tự chọn tham số trên dev (cùng lưới, có touch_top = 100% = không lọc touch, ≥ `min_trades_per_year` lệnh/năm), rồi đo test 2025–2026.

**Tiêu chí:** TB seed tăng và dải seed hẹp lại.

**Cần chạy `train_strategy_h1.ipynb` trước.** Notebook này dùng p_touch từ cache và chỉ train thêm Direction 2 lớp (2 nhãn × các quý × 5 seed, cache ở `artifacts/1h_strategy_v2/folds/dir_*`).

```python
import os, sys
from dataclasses import replace
from pathlib import Path
ROOT = Path.cwd() if (Path.cwd() / "model_training").exists() else Path.cwd().parent
os.chdir(ROOT)
sys.path.insert(0, str(ROOT / "model_training"))
import numpy as np
import pandas as pd
import strategy_h1 as st

pd.set_option("display.width", 220, "display.max_columns", 40)
cfg = st.load_locked("artifacts/1h_strategy_v2/config_locked.json")
OUT = Path(cfg.output_dir) / "exp_direction_2class"
OUT.mkdir(parents=True, exist_ok=True)
D = st.build(cfg)
P_dev, _ = st.walk_forward(D, cfg, "dev")       # cache: touch + Direction 3 lớp
P_test, _ = st.walk_forward(D, cfg, "test")
```

## 1. Train Direction 2 lớp (giữ nguyên p_touch)

```python
P = {"3class": (P_dev, P_test)}
for lab in ("ret8", "tb_sign"):
    c = replace(cfg, direction_label=lab)
    P[lab] = (st.walk_forward_direction(D, c, P_dev, "dev"), st.walk_forward_direction(D, c, P_test, "test"))
```

```python
# AUC: hướng chạm trước (giờ đã chạm) và hướng 8h (mọi giờ)
rows = []
for lab, (pd_, pt_) in P.items():
    for ph, PP in (("dev", pd_), ("test", pt_)):
        a = st.direction_auc(D, PP)
        rows.append({"nhãn": lab, "phase": ph, "AUC chạm trước": a["AUC chạm trước (đã chạm)"].mean(),
                     "AUC hướng 8h": a["AUC hướng 8h (mọi giờ)"].mean(),
                     "quý kém nhất (chạm trước)": a["AUC chạm trước (đã chạm)"].min()})
AUC = pd.DataFrame(rows)
AUC.to_csv(OUT / "auc.csv", index=False)
AUC.round(4)
```

## 2. Mỗi tổ hợp nhãn × thứ tự lọc tự chọn tham số trên dev, đo test

```python
def row(name, c, P_d, P_t):
    dev, test = st.evaluate(D, P_d, c, "dev"), st.evaluate(D, P_t, c, "test")
    S = st.seed_table(D, P_t, c, "test")
    s, sd = test["summary"], dev["summary"]
    yrs = sorted(s["trades_by_year"])
    return {"tổ hợp": name, "touch_top": c.touch_top, "dir_top": c.dir_top, "max_pos": c.max_positions,
            "gap": c.min_gap_bars, "vào": c.entry_mode,
            "dev lệnh/năm": np.mean(list(sd["trades_by_year"].values())), "dev R/lệnh": sd["R_per_trade"],
            "test lệnh": s["trades"], "test R ens": s["R_per_trade"],
            **{f"R {y}": s["R_by_year"][y] / s["trades_by_year"][y] for y in yrs},
            "TB seed": S.loc["TB seed", "R/lệnh"], "std seed": S.loc["std seed", "R/lệnh"],
            "seed tệ nhất": S.loc["seed tệ nhất", "R/lệnh"], "thắng": s["win_rate"], "hết giờ": s["timeout_share"],
            "DD %": s.get("max_dd_pct"), "quý âm": s["negative_quarters"], "LONG": s["long_share"]}, test

rows, res, grids, cfgs = [], {}, {}, {}
for lab in P:
    for go in ("touch_first", "dir_first"):
        name = f"{lab} | {go}"
        base = replace(cfg, direction_label=lab, gate_order=go, locked_from="")
        G = st.grid_search(D, P[lab][0], base)
        G.to_csv(OUT / f"grid_dev_{lab}_{go}.csv", index=False)
        c = st.lock_config(base, st.select_best(G), G, path=str(OUT / f"config_locked_{lab}_{go}.json"))
        r, res[name] = row(name, c, *P[lab])
        rows.append(r); grids[name], cfgs[name] = G, c
B = pd.DataFrame(rows).set_index("tổ hợp")
B.to_csv(OUT / "reselected.csv")
B.round(3)
```

```python
# R/lệnh test theo quý
pd.DataFrame({k: v["by_quarter"]["R/lệnh"] for k, v in res.items()}).round(3)
```

## 3. Touch có giúp khi đã chọn hướng trước không?

Với mỗi tổ hợp, giữ thứ tự lọc và nhãn, lấy ô dev tốt nhất (theo điểm vùng) ở từng mức touch_top. 100% nghĩa là không lọc touch. Sau đó đo test.

```python
rows = []
for name, G in grids.items():
    for tt in sorted(G.touch_top.unique()):
        V = G[(G.touch_top == tt) & G["hợp lệ"] & G["điểm vùng"].notna()]
        if V.empty:
            continue
        best = V.sort_values(["điểm vùng", "điểm"], ascending=False).iloc[0]
        c = replace(cfgs[name], **{a: (best[a].item() if hasattr(best[a], "item") else best[a]) for a in st.GRID})
        r, _ = row(name, c, *P[name.split(" | ")[0]])
        rows.append({"tổ hợp": name, "touch_top": tt, "dev điểm vùng": best["điểm vùng"],
                     **{k: r[k] for k in ("dir_top", "dev lệnh/năm", "test lệnh", "test R ens", "TB seed", "std seed",
                                          "seed tệ nhất", "hết giờ", "DD %")}})
TT = pd.DataFrame(rows)
TT.to_csv(OUT / "by_touch_top.csv", index=False)
TT.round(3)
```

## 4. Chẩn đoán: trong các tín hiệu chỉ có Direction (chưa giới hạn số lệnh), R theo ngũ phân vị p_touch

Dùng dir_top của config đang khoá, `dir_first`, không lọc touch. Nếu R tăng theo p_touch ngoài mẫu thì touch lọc được biến động có ích.

```python
import importlib
st = importlib.reload(st)      # nạp lại strategy_h1 mới (có cột % giá) nếu kernel đang giữ bản cũ
rows = []
for lab, (pd_, pt_) in P.items():
    c0 = replace(cfg, direction_label=lab, gate_order="dir_first", touch_top=1.0)
    for ph, PP in (("dev", pd_), ("test", pt_)):
        S = st.evaluate(D, PP, c0, ph)["all_signal_trades"]
        q = pd.qcut(S.p_touch, 5, labels=[1, 2, 3, 4, 5])
        g = S.groupby(q, observed=True)
        rows.append(pd.DataFrame({"nhãn": lab, "phase": ph, "tín hiệu": g.size(), "R/lệnh": g.R.mean(),
                                  "thắng": g.R.apply(lambda r: (r > 0).mean()), "hết giờ": g.kind.apply(lambda k: (k == 3).mean()),
                                  "chạm thật": g.y_touch.mean()}).rename_axis("nhóm p_touch").reset_index())
Q = pd.concat(rows, ignore_index=True)
Q.to_csv(OUT / "quintiles_p_touch.csv", index=False)
Q.round(3)
```
